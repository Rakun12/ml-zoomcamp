```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import roc_curve
from sklearn.metrics import auc

from sklearn.model_selection import KFold
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction import DictVectorizer
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('data/bank_full.csv', delimiter=';')
df.head()
```

```python
df.columns.values
```

```python
features = ['age', 'job', 'marital', 'education', 'balance',
            'housing', 'contact', 'day', 'month', 'duration',
            'campaign', 'pdays', 'previous', 'poutcome', 'y']
```

```python
df_used = df[features]
```

```python
df_used.y
```

```python
# Changing value for y into binaries
df_used.y = (df_used.y == 'yes')
```

```python
# Splitting data  into train, val, and test
df_full_train, df_test = train_test_split(df_used, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=1)
```

```python
# Reset the index
df_train = df_train.reset_index()
df_val = df_val.reset_index()
df_test = df_test.reset_index()
```

## Q1: ROC AUC feature importance

```python
df_train.dtypes[df_train.dtypes == 'int64']
```

```python
def train(X, y, C=1.0):
    train_dict = X.to_dict()

    dv = DictVectorizer(sparse=False)
    X_train = dv.fit_transform(train_dict)[0]
    
    model = LogisticRegression(C=C, max_iter=1000)
    model.fit(X_train.reshape(-1, 1), y)
    return dv, model


def predict_auc(dv, model, X, y):
    test_dict = X.to_dict()

    X_test = dv.fit_transform(test_dict)[0]
    
    y_pred = model.predict_proba(X_test.reshape(-1, 1))[:, 1]
    fpr, tpr, threshold = roc_curve(y, y_pred)
    auc_score = auc(fpr, tpr)
    return auc_score
```

```python
numeric = ['age', 'balance', 'day', 'duration', 'campaign',
          'pdays', 'previous']
```

```python
print('AUC SCORE')
for item in numeric:
    dv, model = train(df_train[item], df_train.y)
    auc_score = predict_auc(dv, model, df_train[item], df_train.y)
    print(f'{item}: {auc_score.round(3)}')
```

## Q2: Training the model

```python
features_train = ['age', 'job', 'marital', 'education', 'balance',
                  'housing', 'contact', 'day', 'month', 'duration', 
                  'campaign', 'pdays', 'previous', 'poutcome']
```

```python
def training(df, C=1.0):
    X_dict = df[features_train].to_dict(orient='records')

    dv = DictVectorizer(sparse=False)
    X_train = dv.fit_transform(X_dict)

    clf = LogisticRegression(solver='liblinear', C=C, max_iter=1000)
    clf.fit(X_train, df.y)

    return dv, clf
```

```python
def predicting(dv, clf, df):
    X_dict = df[features].to_dict(orient='records')
    
    X_test = dv.transform(X_dict)

    y_pred = clf.predict_proba(X_test)[:, 1]

    return y_pred
```

```python
def auc_score(y, y_pred):
    fpr, tpr, threshold = roc_curve(y, y_pred)
    score = auc(fpr, tpr)
    return score
```

```python
dv, model = training(df_train)
y_pred = predicting(dv, model, df_val)
score = auc_score(df_val.y, y_pred)

score
```

## Q3: Precision and Recall

```python
def prec_recall(y_val, y_pred, total_number):
    scores = []

    threshold = np.linspace(0, 1, total_number)
    
    for t in threshold:
        actual_positive = (y_val == 1)
        actual_negative = (y_val == 0)
    
        predict_positive = (y_pred >= t)
        predict_negative = (y_pred < t)
    
        tp = (predict_positive & actual_positive).sum()
        tn = (predict_negative & actual_negative).sum()
        fp = (predict_positive & actual_negative).sum()
        fn = (predict_negative & actual_positive).sum()
    
        scores.append((t, tp, fp, fn, tn))

    columns = ['threshold', 'tp', 'fp', 'fn', 'tn']
    df_scores = pd.DataFrame(scores, columns=columns)
    
    df_scores['precision'] = df_scores.tp / (df_scores.tp + df_scores.fp)
    df_scores['recall'] = df_scores.tp / (df_scores.tp + df_scores.fn)

    return df_scores
```

```python
df_prec_recall = prec_recall(df_val.y, y_pred, 1001)
df_prec_recall
```

```python
plt.plot(df_prec_recall.precision, df_prec_recall.threshold, label='precision')
plt.plot(df_prec_recall.recall, df_prec_recall.threshold, label='recall')
plt.legend()
```

```python
df_prec_recall[df_prec_recall.precision == df_prec_recall.recall]
```

The intersect of recall and precision is in `threshold` `265` or `266`

## Q4: F1 Score

```python
df_f1_score = prec_recall(df_val.y, y_pred, 101)
df_f1_score
```

```python
df_f1_score['f1_score'] = 2 * (df_f1_score.precision * df_f1_score.recall) / (df_f1_score.precision + df_f1_score.recall)
df_f1_score
```

```python
df_f1_score[df_f1_score.f1_score == df_f1_score.f1_score.max()]
```

The maximal `f1_score` is `0.57`, which in `threshold` `0.22`

## Q5: 5-Fold CV

```python
n_splits = 5
scores = []

kfold = KFold(n_splits=n_splits, shuffle=True, random_state=1)

for train_idx, val_idx in kfold.split(df_full_train):
    df_train = df_full_train.iloc[train_idx]
    df_val = df_full_train.iloc[val_idx]

    dv, model = training(df_train, C=1.0)
    y_pred = predicting(dv, model, df_val)

    score = auc_score(df_val.y, y_pred)
    scores.append(score)

np.std(scores).round(3)
```

## Q6: Hyperparameter Tuning

```python
n_splits = 5
result = []

for C in [0.000001, 0.001, 0.01, 0.1, 0.5, 1, 10]:
    scores = []

    kfold = KFold(n_splits=n_splits, shuffle=True, random_state=1)
    
    for train_idx, val_idx in kfold.split(df_full_train):
        df_train = df_full_train.iloc[train_idx]
        df_val = df_full_train.iloc[val_idx]
    
        dv, model = training(df_train, C=C)
        y_pred = predicting(dv, model, df_val)
    
        score = auc_score(df_val.y, y_pred)
        scores.append(score)

    result.append([C, np.mean(scores), np.std(scores)])
    print('%s %.5f +- %.5f' % (C, np.mean(scores), np.std(scores)))
```
